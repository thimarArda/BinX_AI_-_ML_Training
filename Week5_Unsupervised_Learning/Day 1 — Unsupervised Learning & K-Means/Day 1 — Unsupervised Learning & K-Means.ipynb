# Section 1 : Unsupervised Learning

## 1.1 What Is Unsupervised Learning 
> is a type of machine learning where the model works without labelled data. It learns patterns on its own by grouping similar data points or finding hidden structures without any human intervention.

 > <div style="text-align: center;"> <img src="unsupervised_learning.PNG" alt="how decks look " width="500" > </div>
>
>
- **The "*Interpretation*" stage** signifies that the algorithm doesn't have predefined labels or categories for the data. It groups the data based on inherent patterns.
- **An algorithm** represents unsupervised learning process that helps to identify patterns in the data.
- **The processing stage** shows the algorithm working on the data.
- **The output** shows the results of the unsupervised learning process. In this case, the algorithm might have grouped the animals into clusters based on their species (elephants, camels, cows).

## 1.2 Working of Unsupervised Learning

1. **Collect Unlabeled Data:** Gather a raw dataset that contains input features without any target labels or categories (e.g., untagged images of animals).
2. **Select an Algorithm:** Choose an appropriate model based on your objective, such as clustering (K-Means), association rule learning (Apriori), or dimensionality reduction (PCA).
3. **Train on Raw Data:** Feed the unlabeled data into the algorithm so it can scan for inherent similarities, hidden structures, and relationships.
4. **Group or Transform Data:** The model automatically organizes the data into distinct clusters, rules, or compressed features without human guidance (e.g., grouping visually similar animals together).
5. **Interpret and Apply Results:** Analyze the discovered patterns or structures to gain actionable insights, visualize data, detect anomalies, or feed features into downstream tasks.

## 1.3 Supervised vs Unsupervised Learning

| Feature | Supervised Learning | Unsupervised Learning |
| --- | --- | --- |
| **Input Data** | Labeled data (input features with target labels) | Unlabeled data (only input features) |
| **Goal** | Predict outcomes or classify new data based on past examples | Discover hidden patterns, structures, or groupings in data |
| **Task Types** | Classification (e.g., spam detection) & Regression (e.g., house price prediction) | Clustering (e.g., customer segmentation) & Dimensionality Reduction (e.g., PCA) |
| **Feedback Mechanism** | Direct feedback (model compares predictions against known labels) | No feedback (evaluates intrinsic mathematical structure) |
| **Common Algorithms** | Linear Regression, Decision Trees, Support Vector Machines (SVM), Neural Networks | K-Means Clustering, Hierarchical Clustering, PCA, t-SNE |

# Section 2 :   Clustering Algorithms

> Clustering is an unsupervised machine learning technique used to group similar data points together without using labelled data. It helps discover hidden patterns or natural groupings in datasets by placing similar data points into the same cluster.


 > <div style="text-align: center;"> <img src="clustering.PNG" alt="how decks look " width="500" > </div>

### Attributes of Clustering
- Discover the natural grouping or structure in unlabelled data without predefined categories.
- Data points are assigned to clusters based on similarity or distance measures.
- Uses Euclidean distance, cosine similarity or other metrics depending on data type and clustering method.

### Some common clustering algorithms: 

1. [**K-Means Clustering:**](#section3) Partition data into $K$ predefined groups by assigning points to the nearest cluster centroid based on distance.
2. [**Hierarchical Clustering:**](https://www.geeksforgeeks.org/machine-learning/hierarchical-clustering/) Construct a nested tree of clusters (dendrogram) by sequentially merging smaller clusters (agglomerative) or splitting larger ones (divisive).
3. [**Density-Based Clustering (DBSCAN):**](https://www.geeksforgeeks.org/machine-learning/dbscan-clustering-in-ml-density-based-clustering/) Identify clusters in regions of high point density and label isolated points in low-density regions as noise.
4. [**Mean-Shift Clustering:**](https://www.geeksforgeeks.org/machine-learning/ml-mean-shift-clustering/) Locate dense regions iteratively by shifting data points toward the highest density of nearby points (modes).
5. [**Spectral Clustering:**](https://www.geeksforgeeks.org/machine-learning/spectral-clustering/) Map data onto a lower-dimensional graph-based representation and apply clustering to separate connected nodes.

#  Section 3 : K-Means Clustering

## 3.1 Clustering Mechanics

K-Means is an iterative **expectation-maximization (EM)** algorithm designed to partition $n$ data points into $k$ distinct clusters.

1. **Initialization:** Selecting $k$ centroids. While pure random assignment exists, implementations like `scikit-learn` use **K-Means++** by default, which places initial centroids far apart to accelerate convergence and avoid sub-optimal local minima.
2. **Assignment Step (Expectation):** Every data point $x_i$ computes its Euclidean distance to all $k$ centroids and gets assigned to the nearest one.
3. **Update Step (Maximization):** Each centroid's position is recomputed as the arithmetic mean vector of all data points currently assigned to its cluster.
4. **Convergence:** The two-step cycle repeats until centroid coordinates cease changing (or fall within a predefined tolerance parameter).

--------------

## 3.2 Mathematical Formulation

1. **Euclidean Distance**
The distance between a data point $x$ and a centroid $c$ is calculated as:
$$d(x, c) = \sqrt{\sum_{i=1}^{n} (x_i - c_i)^2}$$



* $x_i$ is the $i$-th feature of the data point.
* $c_i$ is the $i$-th feature of the centroid.
* $n$ is the total number of features.

2. **Centroid Update**
After assigning all data points to clusters, each centroid is updated to the mean of all points assigned to that cluster:
$$c_j = \frac{1}{\vert{}C_j\vert{}} \sum_{x_i \in C_j} x_i$$



* $C_j$ is the set of data points assigned to the $j$-th cluster.
* $\vert{}C_j\vert{}$ is the total number of points in cluster $C_j$.

### 3.4 Choosing $k$ : The Elbow Method

> The Elbow Method helps us decide how many clusters (K) we should use in K-Means
> 
### Detailed Mechanics:
Inertia measures **Within-Cluster Sum of Squares (WCSS)**:


$$\text{Inertia} = \sum_{i=1}^{n} \min_{c_j \in C} (\Vert{}x_i - c_j\Vert{}^2)$$


As $k$ increases toward $n$, inertia naturally decreases toward $0$ (where every point is its own cluster). The Elbow Method plots inertia versus $k$ to locate the point of diminishing marginal returns—where adding another cluster yields a drastically smaller drop in WCSS.

-----

### The graph shows:

 > <div style="text-align: center;"> <img src="1682277078758.png" alt="how elbow plot looks " width="400" > </div>
>
- **X-axis:** the number of clusters, K
- **Y-axis:** inertia (WCSS), which measures how far the data points are from the centers of their clusters.

A lower inertia means the points are more tightly grouped. However, increasing K will always reduce inertia, so we don't simply choose the K with the lowest value. Instead, **we look for the point where increasing K stops giving us a big improvement. This point is called the elbow.**



### Code Syntax Explanation:

```python
# Initialize empty container to store WCSS/inertia for each k candidate
inertias = []

# Loop through candidate k values from 1 to 10
for k in range(1, 11):
    # Fit KMeans model for current value of k
    km = KMeans(n_clusters=k, random_state=42, n_init=10).fit(X)

    # Extract the inertia attribute and append to our list
    inertias.append(km.inertia_)

# Plot k on x-axis vs inertia on y-axis with point markers
plt.plot(range(1, 11), inertias, marker="o")

```

## 3.5 Choosing $k$: The Silhouette Score

**Detailed Mechanics:**
The Silhouette Coefficient $s(i)$ for a point $i$ quantifies how well-clustered it is by comparing cohesion to separation:


$$s(i) = \frac{b(i) - a(i)}{\max(a(i), b(i))}$$

* $a(i)$: Mean distance between $i$ and all other points in its own cluster (intra-cluster cohesion).
* $b(i)$: Mean distance between $i$ and all points in the nearest neighboring cluster (inter-cluster separation).

Scores range from **-1 to +1**:

* **$+1$**: Point is far away from neighboring clusters (ideal assignment).
* **$0$**: Point lies near the decision boundary between two clusters.
* **$-1$**: Point is likely placed in the wrong cluster.

**Code Syntax Explanation:**

```python
from sklearn.metrics import silhouette_score

# Evaluates global mean silhouette coefficient across all samples in X given their assigned labels
score = silhouette_score(X, labels)

```

---

## Feature Scaling Requirement

Because Euclidean distance depends heavily on absolute numerical scale:


$$d(x, c) = \sqrt{(x_1 - c_1)^2 + (x_2 - c_2)^2}$$

If Feature 1 ($x_1$) spans $10,000$ to $100,000$ (e.g., annual income) and Feature 2 ($x_2$) spans $18$ to $80$ (e.g., age), differences in $x_1$ will completely swamp differences in $x_2$. Using `StandardScaler` transforms features to have $\mu = 0$ and $\sigma = 1$, ensuring all dimensions contribute equally to distance calculations.

#### Always scale features before clustering !!!

# Section 4 : Hands-On Lab

The `Iris dataset` contains 150 measurements of iris flowers based on four physical features: sepal length, sepal width, petal length, and petal width. It is a classic machine learning dataset used to classify flowers into three distinct biological species.

**Original Classes:**

1. *Iris-Setosa*
2. *Iris-Versicolour*
3. *Iris-Virginica*

**Goal of the Exercise:**

The goal is to test whether an unsupervised K-Means clustering algorithm can naturally discover these three flower species purely from spatial patterns in the measurement data, without receiving any true class labels.

---



## Step 1: Load and scale a provided dataset (numeric features only) with StandardScaler.


````python
from sklearn.datasets import load_iris

dataset = load_iris()
dataset.keys()
print(dataset.DESCR)
````

````python
import pandas as pd
from sklearn.preprocessing import StandardScaler

#y = dataset.target
X_raw = dataset.data

# Put features into a DataFrame and scale
df_X = pd.DataFrame(X_raw, columns=dataset.feature_names)

scaler = StandardScaler()
X_scaled = scaler.fit_transform(df_X)

# Convert back to DataFrame for cluster analysis
X_scaled_df = pd.DataFrame(X_scaled, columns=dataset.feature_names)
````

## Step 2: Run K-Means for k from 1 to 10 and plot inertia to find the elbow.


````python
import matplotlib.pyplot as plt
from sklearn.cluster import KMeans



inertias = []
k_range = range(1, 11)

for k in k_range:
    km = KMeans(n_clusters=k, random_state=42, n_init=10)
    km.fit(X_scaled_df)
    inertias.append(km.inertia_)

# Plotting the Elbow Curve
plt.figure(figsize=(8, 5))
plt.plot(k_range, inertias, marker="o", linestyle="--", color="b")
plt.xlabel("Number of Clusters (k)")
plt.ylabel("Inertia (Within-Cluster Sum of Squares)")
plt.title("Elbow Method For Optimal k")
plt.xticks(k_range)
plt.grid(True)
plt.show()
````

###  Interpretation

* **$k = 1$:** All data points are placed into a single cluster, resulting in maximum inertia (600) as distances are measured from a single general center.
* **$k = 2$:** Inertia drops significantly to 225. Splitting the dataset into two distinct groups drastically increases cluster compactness.
* **$k = 3$:** Inertia improves further, dropping to 140, indicating another meaningful reduction in within-cluster variance.
* **Beyond $k = 3$:** The rate of decrease slows substantially. Adding more clusters past this point yields diminishing returns, as inertia declines at a much flatter rate.

### The Elobow
The graph suggests that the main change happens around K = 2 and 3.
But the Elbow Method alone doesn't give us a final answer, so we are going to compare the two clusters at step 3.

## Step 3: Compute the silhouette score for the top two candidate k values and pick the best

````python
from sklearn.cluster import KMeans
from sklearn.metrics import silhouette_score

# Candidate k values from the Elbow plot
k_candidates = [2, 3]
silhouette_scores = {}

for k in k_candidates:
    # 1. Fit KMeans for candidate k
    km = KMeans(n_clusters=k, random_state=42, n_init=10)
    cluster_labels = km.fit_predict(X_scaled_df)

    # 2. Compute average Silhouette Score
    score = silhouette_score(X_scaled_df, cluster_labels)
    silhouette_scores[k] = score
    print(f"Silhouette Score for k = {k}: {score:.4f}")

# 3. Automatically select the best k
best_k = max(silhouette_scores, key=silhouette_scores.get)
print(
    f"\nOptimal choice: k = {best_k} (Highest Silhouette Score: {silhouette_scores[best_k]:.4f})"
)
````

### Based on the Silhouette Analysis:



The score for **$k = 2$** (**0.5818**) is  higher than  **$k = 3$**, meaning that a two-cluster model achieves better separation across the feature space.
 Although both scores are positive, **$k = 2$** provides a better balance between:
- keeping points close to the other points in their own cluster.
- keeping them far away from points in other clusters.

Therefore, **$K = 2$** is the better choice.

## Step 4: Fit final K-Means with the chosen k, and visualize the clusters on a 2D scatter plot.

> Since the Elbow Method and Silhouette Analysis both led us to select K=2, we fit the final K-Means model using two clusters. The model creates two centroids and assigns each Iris sample to one of the two clusters. We then visualize these two clusters on a 2D scatter plot.

````python
import matplotlib.pyplot as plt
import seaborn as sns
from sklearn.decomposition import PCA

# 1. Fit final K-Means with chosen k = 2
km_final = KMeans(n_clusters=2, random_state=42, n_init=10)
labels = km_final.fit_predict(X_scaled_df)

# 2. Reduce 30 features down to 2 components for 2D visualization
pca = PCA(n_components=2)
X_pca = pca.fit_transform(X_scaled_df)
centroids_pca = pca.transform(km_final.cluster_centers_)

# 3. Scatter plot of data points colored by cluster + centroids
plt.figure(figsize=(8, 6))
sns.scatterplot(
    x=X_pca[:, 0], y=X_pca[:, 1], hue=labels, palette="viridis", s=50
)

# Plot the centroids as large red 'X' markers
plt.scatter(
    centroids_pca[:, 0],
    centroids_pca[:, 1],
    c="red",
    marker="X",
    s=200,
    label="Centroids",
)

plt.xlabel("PCA Component 1")
plt.ylabel("PCA Component 2")
plt.title("K-Means Clusters in 2D Space")
plt.legend()
plt.show()
````

### Plot Analysis : 
- **Cluster 1:** A small, compact group on the left, clearly separated from the other points. This mainly represents `Iris-Setosa`.
- **Cluster 0:** A larger group on the right containing both `Versicolor` and `Virginica`, which have more similar measurements and overlap.

### Copmaring with the true lables


Although Iris has 3 known species, K-Means groups points are based only on similarity and distance, not their actual labels.
`Setosa` is very different, so it forms it's own cluster, while `Versicolor` and `Virginica` are more similar, so K-Means groups them together.

So waht happend was that the clustering performed reasonably well: it successfully identified and separated Iris-Setosa, which has a distinct pattern. However, it could not clearly distinguish between Versicolor and Virginica because their features are more similar and overlapping
