## 6) Outlier Detection with the IQR Method

> Outliers are observations that fall far outside the typical range of the data. They may represent valid but rare cases or data entry errors. Since outliers can affect statistical analysis and machine learning models, they should always be investigated before deciding how to handle them.
>
> 

The **Interquartile Range (IQR)** method is a common technique for detecting outliers.

- **Q1 (First Quartile):** The 25th percentile.
- **Q3 (Third Quartile):** The 75th percentile.
- **IQR:** The difference between Q3 and Q1.

The lower and upper limits are calculated as:

$$
\text{Lower Bound} = Q1 - 1.5 \times IQR
$$

$$
\text{Upper Bound} = Q3 + 1.5 \times IQR
$$

Any value below the lower bound or above the upper bound is considered an **outlier**.

> **Note:** An outlier is not automatically an error. It should be examined before deciding whether to keep, remove, or modify it.

```python
figure, axes = plt.subplots(1, len(numeric_cols), figsize=(15, 5))
```

- `figure` = a variable that holds all the subplots
- `axes` = a list of individual plot "slots" (one per column)
- `zip(axes, numeric_cols)` pairs each slot with a column name, position by position
- `df[col]` = pulls just that column's data for `ax` to plot into
-  `tight_layout()` automatically adjusts subplot parameters so that all subplots fit cleanly within the figure area

````python
numeric_cols = df.select_dtypes(include='number').columns # or: numeric_cols = ['age', 'bmi', 'children', 'charges']

figure, axes = plt.subplots(1, len(numeric_cols), figsize=(15, 5))
for ax, col in zip(axes, numeric_cols):
    sns.histplot(data=df[col], ax=ax)
    ax.set_title(col)
plt.tight_layout()
plt.show()
````

## Step 2: Produce a box plot for the key numeric variables and identify any outliers visually.


````python
numeric_cols = df.select_dtypes(include='number').columns

fig, axes = plt.subplots(1, len(numeric_cols), figsize=(4 * len(numeric_cols), 5))

for ax, col in zip(axes, numeric_cols):
    sns.boxplot(y=df[col], ax=ax)
    ax.set_title(col)

plt.tight_layout()
plt.show()
````

> The box plots show visible outliers in the **bmi** and **charges** variables. The bmi column has a few unusually high values. The charges column also has several high-value outliers, which means some individuals have much higher medical costs than the rest of the dataset.

## Step 3: Apply the IQR method in code to flag outliers in at least one column, and decide (with justification) how to handle them

````python
Q1 = df["bmi"].quantile(0.25)
Q3 = df["bmi"].quantile(0.75)
IQR = Q3 - Q1
outliers = df[(df["bmi"] < Q1 - 1.5*IQR) | (df["bmi"] > Q3 + 1.5*IQR)]
print(outliers)
````

 ### Outlier Detection

The IQR method was applied to the **bmi** column to identify outliers. It detected **9** observations with unusually high bmi values.

However , I think these outliers should be kept because they appear to be valid records rather than data entry errors. Removing them could result in the loss of important information, especially since people with very high bmi values may have higher medical costs. Therefore, the outliers were retained for further analysis.


## Step 4: Produce a count plot for each categorical variable and note any class imbalance.

````python
categorical_cols = ['sex', 'smoker', 'region']

fig, axes = plt.subplots(1, len(categorical_cols), figsize=(5 * len(categorical_cols), 5))

for ax, col in zip(axes, categorical_cols):
    sns.countplot(data=df, x=col, ax=ax)
    ax.set_title(f'Count of {col}')

plt.tight_layout()
plt.show()
````
