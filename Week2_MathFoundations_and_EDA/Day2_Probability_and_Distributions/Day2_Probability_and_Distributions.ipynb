## What is Probability, and Why Does AI Need It?

Probability is the way of measuring how likely something is to happen, using a
number between 0 (impossible) and 1 (certain).

AI models rarely give a plain yes-or-no answer. Instead, they usually give an answer
like "there is an 85% chance this patient has heart disease" or "there is a 70% chance
this email is spam." That confidence number is probability at work.

Pprobability lets an AI model:
- Handle uncertainty instead of pretending everything is 100% certain
- Update its beliefs as new information (data) comes in
- Rank how confident it is in each possible answer, instead of only picking one

### 1) Probability Basics



> <span style="font-size: 20px;"> **Defenition :** Probability is defined as the likelihood of the occurrence of any event. It gives a numerical value to the chance or likelihood of something happening.</span>

$P(E) = Number of Favorable Outcomes  /  Total Number of Outcomes$

The probability of an event $E$ , denoted by $P(E)$, is a number between 0 and 1 that represents the likelihood of $E$ occurring.

- If $P(E) = 0$ , the event $E$  is impossible.
- If $P(E) = 1$ , the event $E$  is certain to occur.
- If  $0 < P(E) < 1$, the event $E$  is possible but not guaranteed.

### 2) Core Probability Rules

#### Addition Rule:

 >  $P(A or B)$      or        $P(A∪B) = P(A) + P(B) - P(A∩B)$,
   where A∪B denotes the union of events A and B.
>
> Applied when calculating the occurrence of **multible events together** (The chance of either event)

##### Example:
 In a survey of 1000 students, it was found that 600 prefer the BinX Online training batch, 300 prefer the Offline training batch, and 100 like both the Online and Offline training batches. What is the probability that a randomly selected student from this survey likes either the Online or Offline training batch?

##### Solution:
Total number of students : 1000
Probability to like the online training : $P(A) = 600/1000$

Probability to like the offline training : $P(B) = 300/1000$

Probability to like the both ofline and offlibe training :  $100/1000$


a randomly selected student from this survey likes either the Online or Offline training batch : $600/1000 + 300/1000 -100/1000 = 0.8$

#### Multiplication Rule 
> $P(A∩B) = P(A) × P(B)$, where A and B are independent events.
>
> for **Independent Events occurring one after the other**

##### Example:
A vessel contains 4 blue balls, 5 red balls, and 11 white balls. If three balls are drawn from the vessel at random, what is the probability that the first ball is red, the second ball is blue, and the third ball is white?

##### Solution:
Probability to get the first ball is red or the first event is 5/20

Since we have drawn a ball for the first event to occur, then the number of possibilities left for the second event to occur is 20 – 1 = 19

Hence, the probability of getting the second ball as blue or the second event is 4/19

Again with the first and second event occurring, the number of possibilities left for the third event to occur is 19 – 1 = 18

And the probability of the third ball is white or the third event is 11/18

Therefore, the probability is 5/20 x 4/19 x 11/18 = 220/6840 = 11/342 = 0.032 

We can express it as: P = 3.2%.

#### Complement Rule:
> $ P(A ′) = 1 - P(A)$, where ′ A ′ denotes the complement of event A.
>
> **The chance something does NOT happen**

##### Example:
 A student has a 75% chance of passing an exam. What is the probability that the student will **not** pass the exam?

##### Solution:
P(A) = 75% 

The chance of the student to fail is : 1- 0.75 = 0.25

### 4) Bayes' Theorem

#### Proof

Previously, we know that:

$$
P(A \mid B) = \frac{P(A \cap B)}{P(B)}
$$

Also,

$$
P(B \mid A) = \frac{P(B \cap A)}{P(A)}
$$

Since

$$
P(A \cap B) = P(B \cap A),
$$

we can write:

$$
P(B \mid A) = \frac{P(A \cap B)}{P(A)}
$$

Multiplying both sides by $P(A)$:

$$
P(B \mid A) \cdot P(A) = P(A \cap B)
$$

Substitute this into the first equation:

$$
P(A \mid B)
= \frac{P(A \cap B)}{P(B)}
= \frac{P(B \mid A)\cdot P(A)}{P(B)}
$$

Therefore, **Bayes' Theorem** is:

$$
P(A \mid B)
=
\frac{P(B \mid A)\cdot P(A)}{P(B)}
$$

---

Similarly, we can rearrange the formula to obtain:

$$
P(B \mid A)
=
\frac{P(A \mid B)\cdot P(B)}{P(A)}
$$

### 5) Common Distribution

 #### 1) Normal Distribution

> A normal distribution, often called a **_bell curve_** or **_Gaussian distribution_**, is a continuous probability distribution **where data is symmetrically clustered around the center**. In this distribution, the **mean, median, and mode are equal**, and the **majority of values lie near the mean**, tapering off symmetrically toward the extremes
>
> 
>
> **Note :** The normal distribution matters most: many statistical methods and ML assumptions rely on data being approximately
normal, which is one of the first things EDA checks for.


##### Structure:
 - Looks like a smooth, even bell-shaped wave.
 -  It measures continuous data (like exact heights, weights, or test scores) where any decimal value is possible.

#### 2) Binomial Distribution

> Binomial distribution is a probability distribution used to model the number of successes in a fixed number of independent trials, where each trial has only two possible outcomes: success or failure.
>
**Example :** Flipping a coin 3 times
#### Core Requirements (BINS)
- **Binary outcomes:** Each trial results in only two possibilities, typically called "success" or "failure".
- **Independent trials:** The result of one trial does not change or affect the result of any other trial.
- **Number of trials:** There is a fixed, pre-determined total number of trials, written as n.
- **Success probability:** The chance of success, written as p, stays exactly the same for every single tria

```python
import numpy as np
import matplotlib.pyplot as plt
```

### Step 1: Simulate 10,000 coin flips with NumPy and confirm the proportion of heads approaches 0.5.

### Method 1 :Using a loop

```python
number_of_flips = 10000
head_value = 1
tale_value = 0

val = np.random.choice([tale_value , head_value], size=number_of_flips)
count = 0
for x in val:
    if   x ==1:
        count+=1
proportion = count / number_of_flips
print(" Head probability ", proportion)
    
```

#### Method 2 :
Since 0 does not affect the average and only 1 values contribute to the sum, we can use np.mean() to calculate the proportion of heads directly. The average of the list represents the probability of getting heads.

```python
val = np.random.choice([0, 1], size=10000)

proportion = np.mean(val)
print("Head probability:", proportion)
```

### Step 2: Sample from a normal distribution with np.random.normal and plot its histogram to confirm the bell shape

The `np.random.normal()` function in NumPy draws random samples from a normal (Gaussian) distribution
> - `loc` (float): The mean ("centre") of the distribution.
> - `scale` (float):
>  > -  The standard deviation (spread or width).
>  > -   Must be non-negative.
>  > -   Default is 1.0.
> -  `size` (int or tuple of ints): The output shape. Returns a single value if `None`

```python
data = np.random.normal(loc=0.0, scale=2.0, size=10000)
plt.hist(data, bins=30)

plt.title("Histogram of a Normal Distribution")
plt.xlabel("Value")
plt.ylabel("Frequency")
plt.show()
```

<span style="font-size:120%"> The function creates distributed values around the center of the specified mean, which is zero in this case. It results in a normal distribution where most of the values are centered around the mean, and fewer values are farther away from it. This leads to a bell-shaped curve.</span>.
