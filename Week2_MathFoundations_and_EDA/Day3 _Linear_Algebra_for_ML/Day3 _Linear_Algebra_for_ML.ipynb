### Vectorization

#### Why are vectors so important in Machine Learning?
>Vectors allow us to represent data in a mathematical form that computers can process efficiently.
Instead of processing one value at a time, we can process an entire vector of values at once. This idea is called **vectorization.**

#### Vectorization 
>is the process of performing the same operation on an entire vector (or matrix) instead of looping through each element individually

#### For example, suppose we have the ages of five people:

````python
ages = [20, 25, 18, 30, 22]
````

#### If we want to add one year to everyone's age, we could use a loop:

````python
new_ages = []

for age in ages:
    new_ages.append(age + 1)
````

#### With NumPy, we can treat the whole list as a vector:

````python
import numpy as np

ages = np.array([20, 25, 18, 30, 22])
new_ages = ages + 1

#  result : [21 26 19 31 23]
````

#### Why Vectorization?
Machine Learning models often work with **millions** of numbers.

Instead of processing every pixel one by one, we represent all of them as vectors or matrices and apply mathematical operations to the whole dataset simultaneously.

This makes Machine Learning:

- Much faster.
- More memory efficient.
- Easier to write and understand.
- Able to take advantage of optimized libraries like NumPy, which are implemented in low-level languages such as C.

A matrix used for tabular data has **two dimensions**:

- **Rows (\(m\))**
- **Columns (\(n\))**

Its size is written as an **\(m \times n\)** matrix, meaning:

- **\(m\)** = number of rows (samples)
- **\(n\)** = number of columns (features)

### Structure of Tabular Matrices

- **Rows (\(m\))**: Represent individual items, records, or samples (horizontal lines).
- **Columns (\(n\))**: Represent variables, features, or attributes (vertical lines).
- **Order**: Matrix dimensions are always written as **rows first**, followed by **columns**.

## 3) The dot product and how models predict with it

### What is  Dot Product

The **dot product** is an operation between two vectors. It multiplies their corresponding elements and then adds the results together, producing a **single scalar value**.


### Mathematically:

$$
\mathbf{a} \cdot \mathbf{b} = \sum_{i=1}^{n} a_i b_i
$$

For example:

```python
import numpy as np

features = np.array([25, 50000, 3])
weights = np.array([0.1, 0.0002, 1.5])

prediction = np.dot(features, weights)

print(prediction)
```

Output:

```text
17.0
```

The calculation is:

$$
(25 \times 0.1) + (50000 \times 0.0002) + (3 \times 1.5)
$$

$$
= 2.5 + 10 + 4.5 = 17.0
$$


### Why is the dot product important in Machine Learning?

The dot product is one of the **most fundamental operations** in Machine Learning. It combines the input features with their corresponding weights to produce a prediction.

For a linear model, the prediction is computed as:

$$
\text{prediction} = \mathbf{x} \cdot \mathbf{w} + b
$$

where:

- **$\mathbf{x}$** is the feature vector.
- **$\mathbf{w}$** is the weight vector learned by the model.
- **$b$** is the bias term.

This is exactly how models such as **Linear Regression** and **Logistic Regression** compute their output before making a prediction.

````python
features = np.array([25, 50000, 3])
weights = np.array([0.1, 0.0002, 1.5])
prediction = np.dot(features, weights) # 2.5 + 10 + 4.5 = 17.0
````

### Summary

> The dot product measures the weighted sum of the input features. Since many Machine Learning models rely on the equation
>
> $$
> \text{prediction} = \mathbf{x} \cdot \mathbf{w} + b,
> $$
>
> understanding the dot product is essential for understanding how these models make predictions

## 4) Matrix multiplication and the shape-matching rule


**Matrix multiplication** extends the idea of the dot product to entire datasets. Instead of computing the prediction for one sample at a time, it computes predictions for **all samples simultaneously**.

For matrix multiplication, the dimensions must follow this rule:

$$
(m \times n) \times (n \times p) = (m \times p)
$$

The **inner dimensions** must match. If they do not, matrix multiplication is not possible.

For example:

- A matrix of shape **(2, 3)** can be multiplied by a matrix or vector of shape **(3, 1)** or **(3,)**.
- The result will have shape **(2, 1)** or **(2,)**, meaning one output for each sample.

### Example, with code

````python
import numpy as np

# Feature matrix (2 samples, 3 features)
X = np.array([
    [25, 50000, 3],
    [40, 80000, 10]
])

# Weight vector (3 features)
w = np.array([0.1, 0.0002, 1.5])

# Compute predictions for all samples
predictions = X @ w

print(predictions)
````


Output:
```text
[17.0 33.5]
```

The shapes are:
- **Feature matrix:** `(2, 3)`
- **Weight vector:** `(3,)`
- **Output predictions:** `(2,)`

Each row of `X` is multiplied with the weight vector using the **dot product**, producing one prediction for each sample.


### Why is matrix multiplication important in Machine Learning?

Machine Learning models are trained on datasets containing **thousands or even millions of samples**. Computing predictions one sample at a time would be inefficient.

Matrix multiplication allows us to:

- Compute predictions for the entire dataset in a single operation.
- Take advantage of optimized numerical libraries such as **NumPy**.
- Significantly improve speed and efficiency through vectorized computations.




### Understanding Shapes

One of the most common errors in Machine Learning is a **shape mismatch**.

For matrix multiplication to work:

$$
(m \times n) \times (n \times p)
$$

the **inner dimensions** must be equal.

For example:

| Left Matrix | Right Matrix | Valid? | Result Shape |
|-------------|--------------|:------:|--------------|
| `(2, 3)` | `(3,)` | Yes | `(2,)` |
| `(100, 20)` | `(20,)` | Yes | `(100,)` |
| `(5, 4)` | `(4, 2)` | Yes | `(5, 2)` |
| `(2, 3)` | `(2,)` | No | Shape mismatch |


### Summary

> Matrix multiplication applies the **dot product** to every row of a dataset at once, allowing Machine Learning models to generate predictions for all samples efficiently. Understanding matrix shapes is essential because matrix multiplication only works when the inner dimensions match.

````python
import numpy as np

# Feature matrix (4 samples, 4 features)
X = np.array([
    [2, 3, 4, 5],
    [1, 6, 2, 8],
    [4, 2, 7, 3],
    [4, 2, 7, 3]])

# weight vector (3 weights)
w = np.array([4, 5, 6])

predictions = X @ w
print(predictions)
````
