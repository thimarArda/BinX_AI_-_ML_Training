# Section 2 : Why Deep Learning

Classical ML (Weeks 3-4) works well on **structured, tabular data** — rows and columns with clearly
defined features. Deep learning shines where classical methods struggle: **images, text, audio**, and
other high-dimensional, unstructured data with complex patterns.

The key difference: a neural network learns its own useful features directly from raw data, instead of
relying on features engineered by hand (like the `FamilySize` or `Deck` features built manually in Week
4). This is exactly why neural networks power modern computer vision and NLP — the topics of Weeks 7-8,
and directly relevant to this project's sentiment analysis text data.

# Section 3 : The Neuron

## 3.1 What a Single Neuron Does

A single neuron is the building block of a neural network, and it is **exactly the dot product operation
from Week 2**, with two small additions:

1. Take the inputs and multiply each one by a weight.
2. Sum all of those weighted inputs together — this is the dot product from Week 2.
3. Add a **bias** term.
4. Pass the result through an **activation function**.

In one line:

$$
\text{output} = \text{activation}\big(\mathbf{w} \cdot \mathbf{x} + b\big)
\qquad\text{where}\qquad
\mathbf{w} \cdot \mathbf{x} = w_1 x_1 + w_2 x_2 + \dots + w_n x_n
$$

A neural network is just **many of these neurons, connected in layers**.


## 3.2 Implementing a Single Neuron

```python
import numpy as np

def sigmoid(z):
    # A common activation function - squashes any real number into the (0, 1) range
    return 1 / (1 + np.exp(-z))

def neuron(inputs, weights, bias, activation=sigmoid):
    # Step 1 & 2: weighted sum - this IS the dot product from Week 2
    z = np.dot(inputs, weights) + bias
    # Step 3: activation
    return activation(z)

# Example: a neuron with 3 inputs
x = np.array([0.5, -1.2, 3.0])      # inputs
w = np.array([0.4, 0.7, -0.3])      # weights (learned parameters)
b = 0.1                             # bias (also learned)

output = neuron(x, w, b)
print(f"z (weighted sum + bias) = {np.dot(x, w) + b:.4f}")
print(f"neuron output (after activation) = {output:.4f}")
```

**Syntax explained:**
- `np.dot(inputs, weights)` — the dot product from Week 2: multiplies each input by its corresponding
  weight and sums the results into a single number.
- `+ bias` — a constant added on top, letting the neuron shift its output up or down independent of the
  inputs (the same role an intercept plays in linear/logistic regression).
- `sigmoid(z)` — one common activation function; it squashes the raw weighted sum into a bounded range
  `(0, 1)`, which is why it's often used when the output should look like a probability. Other activation
  functions (ReLU, tanh, etc.) shape this differently, but every one of them takes the same `z` as input.


````python
import numpy as np

def sigmoid(z):
    # A common activation function - squashes any real number into the (0, 1) range
    return 1 / (1 + np.exp(-z))

def neuron(inputs, weights, bias, activation=sigmoid):
    # Step 1 & 2: weighted sum - dot product 
    z = np.dot(inputs, weights) + bias
    # Step 3: activation
    return activation(z)

# Example: a neuron with 3 inputs
x = np.array([0.5, -1.2, 3.0])      # inputs
w = np.array([0.4, 0.7, -0.3])      # weights (learned parameters)
b = 0.1                             # bias (also learned)

output = neuron(x, w, b)
print(f"z (weighted sum + bias) = {np.dot(x, w) + b:.4f}")
print(f"neuron output (after activation) = {output:.4f}")
````

**Reading the result:** `z` is just a number — it can be any real value, positive or negative,
large or small, depending on the inputs, weights, and bias. The activation function is what turns that raw
number into something more useful — here, a value between 0 and 1 that could be interpreted as "how
confident is this one neuron that its inputs indicate the positive class?"

# Section 4 : Layers and Architecture

Neurons are organized into **layers**, and data flows through them in order:

| Layer | Role |
|---|---|
| **Input layer** | Receives the raw features — one node per feature (e.g. one node per word/token feature for text) |
| **Hidden layer(s)** | Learn intermediate representations; more layers or wider layers = more capacity to learn complex patterns |
| **Output layer** | Produces the final prediction — its shape depends on the task (e.g. 3 output nodes for positive/neutral/negative sentiment) |

**"Deep" learning** simply means a network with **multiple hidden layers**. Each layer transforms its
input and passes the result forward to the next layer, and stacking layers lets the network build
increasingly abstract patterns — early layers might respond to simple patterns, while deeper layers
combine those into more complex, task-specific representations.


### A Small Visual

* **Shallow Neural Network:** A basic machine learning model with only **one hidden layer** between the input and output. It handles simple patterns and small datasets.
* **Deep Neural Network:** A complex machine learning model with **two or more hidden layers** (often dozens or hundreds). It can learn intricate, high-level patterns from massive amounts of data like images or voice.

# Section 5 : Weights and Biases Are What's Learned

The **weights** and **biases** are the network's parameters — the numbers that actually get adjusted
during training. A freshly created network starts with random weights and makes essentially random
predictions; training is the process of nudging these weights, little by little, until the predictions
become accurate.



> **The core idea:** the neuron is  dot product, plus a bias,
> plus an activation function. A neural network is that same operation, stacked and repeated many
> times over.
