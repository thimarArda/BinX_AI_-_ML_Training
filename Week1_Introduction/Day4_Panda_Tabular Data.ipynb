## **2) Cleaning Data**

### Handling Missing Values and Duplicate Data

Before performing data analysis, it is important to check and clean the dataset to ensure data quality. Missing values and duplicate records can affect the accuracy of analysis results, so the following pandas functions are used for data cleaning:

#### 1. Checking for Missing Values


`df.isnull().sum()`

>This function checks each column in the dataset and returns the number of missing values (NaN) present in each column. It helps identify whether any columns require further cleaning.

```python
df.isnull().sum() # The function returned 0 meaning there werent any missing values in the columns
```



#### 2. Filling Missing Values


`df["col_name"].fillna(df["col_name"].mean(), inplace=True)`

>This function replaces missing values in a selected column with a specified value. In this example, missing values in the  column are replaced with the mean column value. This technique keeps the existing rows while preventing missing values from affecting calculations and analysis.



#### 3. Removing Rows with Missing Values


`df.dropna(inplace=True)`

>This function removes any rows that contain missing values. It is useful when missing values cannot be replaced or when removing incomplete records is more appropriate than filling them.



#### 4. Removing Duplicate Records


`df.drop_duplicates(inplace=True)`

>This function identifies and removes duplicate rows from the dataset. Removing duplicates ensures that repeated records do not affect statistical calculations, visualizations, or machine learning models.

## **4) Performing Grouping and Aggregation On the Data**

- We use the function `groupby()` to group the data into catagories.
- The `groupby()` operation is divided into two main steps: 
  - **Step 1:** Splitting data into groups.
  - **Step 2:** Applying Functions to Groups (we will only be using aggregation for now)

 

## Grouping

### 1) Methods for splitting data 

**Method 1:** Group data by a single key
>The `.groups` attribute returns a dictionary where:
 >- Key = the unique values in the column 
 >- Value = lists of the row indices that belong to each unique value. 

```python
df.groupby('floors')
print(df.groupby('floors').groups)   # the output for this code is a dictionary where it has:
                                     # Keys representing the number of floors (e.g., 1.0, 1.5, 2.0, 2.5, 3.0, 3.5).
                                     # Values are lists of row indices corresponding to houses with that number of floors.
```

**Method 2:** Grouping data with multiple keys
>In order to group data with multiple keys, we pass multiple keys in groupby function
 

```python
df.groupby(['waterfront', 'view'])
print(df.groupby(['waterfront', 'view']).groups)
```

**Method 3:** Grouping data by sorting keys
> Syntax : `df.groupby('Column1')['Column2'].sum()`
> - Column1 → used to create the groups.
> - Column2 → the values to be added together.
> - `.sum()` → performs an addition of all the values in Column2 within each group.


```python
df.groupby('floors')['price'].sum()
```

**Method 4:** Iterating through groups:
> Syntax : `df.groupby('Column1')['Column2'].sum()`
> - Column1 → used to create the groups.
> - Column2 → the values to be added together.
> - `.sum()` → performs an addition of all the values in Column2 within each group.


```python
grp = df.groupby('floors')
for name, group in grp:
    print(name)         # prints the catagory og the floor (1.0 ,1.5,.....,3.5)
    print(group)
    print()
```

 ## Aggregation
 >#### Aggregation allows us to find a summary statistic for each group like summing or averaging values.

```python
grp1 = df.groupby('floors')
grp1['waterfront'].aggregate(np.sum)
```

```python
grp = df.groupby('floors')
grp['price'].agg([np.sum, np.mean, np.std])
```
