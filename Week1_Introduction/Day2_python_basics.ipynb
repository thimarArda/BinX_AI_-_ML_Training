## **Step 1: Write a function that takes a list of numbers and returns the mean, min, and max as a dictionary**

### For the first excercise, I wanted to get used to the python syntax. So, instead of using functions from the math library I made the functions myself.
- The first cell contains three functions that calculate the avarage, minimum number, and maximum number of a given list of integers.
- The second cell contains a functions that takes a list of numbers and perform the three calulations by calling the three functions in the previous cell.

```python
numbers =[10, 20, 40,80,60,40]
def min_number(numbers):    #   function that selects the minimum number in the list
     min=numbers[0]
     for num in numbers:
        if num<min:
            min = num
     return min
    
def max_number(numbers):    #   function that selects the maximum number in the list
     max=numbers[0]
     for num in numbers:
        if num>max:
            max = num
     return max

   
def mean_number(numbers):    #   function that calculates the avarage of the numbers in the list
    total=0;
    for num in numbers:
        total += num
    mean= total/len(numbers)
    return mean
```

```python
def list_to_dict(numbers):
    return{
        'mean' : mean_number(numbers),
         'min' : min_number(numbers),
        'max' : max_number(numbers) 
    }
print(list_to_dict(numbers))
```

## **Step 2: Rewrite a given for-loop that filters even numbers as a single list comprehension.**

```python
for value in [1, 2, 3, 4]:
    if value % 2 == 0:
      print(f"{value} is even")
else:
      print(f"{value} is odd")
```

### Performing the previous cell as a one-line syntax using list comprehension

```python
even_numbers = [num for num in [1, 2, 3, 4] if num % 2 == 0]
print(even_numbers)
```

## **Step 3 :Define a small class representing a data record with at least two attributes and one method.**

### The class Age takes a person with a name and age as attributes. It contains a method that determines whether the person is a teenager or not

```python
class Age:
    def __init__(self, name, age):
        self.name = name
        self.age = age

    def is_a_teenager(self):
        if 13 <= self.age <= 19:
            return self.name + " is a teenager"
        else:
            return self.name + " is not a teenager"

person = Age("Thimar", 22)
person2 = Age("Masa", 16)
print(person.is_a_teenager())
print(person2.is_a_teenager())
```
